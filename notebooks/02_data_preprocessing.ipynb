## 1. Load and validate

`load_raw` reads the csv, renames columns, builds a timestamp index from year / month / day / hour and stops
with an error if the column list changes, the hourly grid has holes or duplicates, or any weather value is
missing. These are exactly the properties the EDA found (A1, A2); the checks make sure they still hold if the file
is ever replaced. Only `pm25` is allowed to be missing.

```python
raw = load_raw(cfg["paths"]["raw"])
print(raw.shape, raw.index.min(), "->", raw.index.max())
raw.head()
```

```python
raw.isna().sum().rename("missing").to_frame().T
```

## 2. Decode cumulative wind speed

EDA A5 showed that `Iws` adds up wind speed while the direction stays the same and restarts when `cbwd` changes.
The hourly speed is the difference inside a direction run, and the raw value on the first hour of a run:

$$ws_t = \begin{cases} I_t & d_t \neq d_{t-1} \\ I_t - I_{t-1} & d_t = d_{t-1} \end{cases}$$

```python
df = decode_wind(raw)
df.loc["2010-01-01 12:00":"2010-01-01 18:00", ["cbwd", "ws_cum", "ws"]]
```

Check: adding the hourly speeds back up inside each run rebuilds `Iws` exactly.

```python
run_id = df["cbwd"].ne(df["cbwd"].shift()).cumsum()
np.allclose(df.groupby(run_id)["ws"].cumsum(), df["ws_cum"])
```

```python
df.groupby("cbwd")["ws"].describe().round(2)
```

## 3. Snow and rain flags

`Is` and `Ir` are counters that are above 0 while snow or rain is falling (EDA A5). The flags keep that reliable
part; the counters are kept as well.

```python
df = decode_precip(df)
df[["snow", "rain"]].mean().mul(100).round(2).rename("% of hours").to_frame().T
```

## 4. Wind direction one-hot

Four columns in a fixed order (NE, NW, SE, cv), so column positions never depend on which directions appear in a
fold.

```python
df = encode_dir(df, cfg["clean"]["wind_dirs"])
df.filter(like="wd_").sum().rename("hours").to_frame().T
```

## 5. Missing target: label, flag, causal input

The rule comes from EDA A3: most gaps are short, a value carried forward is off by about 13 after 1 hour and 27
after 3 hours, and interpolation would use the future.

- `pm25` stays untouched. A missing value means that hour is skipped in training loss and evaluation.
- `pm25_miss` records where the value was missing.
- `pm25_in` is the input copy: the last observed value carried forward for at most 3 hours. Longer gaps stay
  missing; XGBoost handles missing inputs natively.

Linear interpolation is shown for comparison only. It fills the gap using the value **after** it, which leaks the
future into inputs near the forecast origin.

```python
df = fill_target(df, cfg["clean"]["ffill_limit"])

g = gaps(df["pm25"].isna())
ex = g[(g["size"] > 3) & (g["size"] <= 6)].iloc[0]
s = ex["start"]
win = df.loc[s - pd.Timedelta("2h"): s + pd.Timedelta(hours=int(ex["size"]) + 1), ["pm25", "pm25_miss", "pm25_in"]]
win.assign(interpolated_leaks=df["pm25"].interpolate(limit_area="inside").loc[win.index].round(1))
```

```python
pd.Series({
    "missing labels": df["pm25"].isna().sum(),
    "missing inputs after fill": df["pm25_in"].isna().sum(),
    "input hours recovered": df["pm25"].isna().sum() - df["pm25_in"].isna().sum(),
}).to_frame("hours").T
```

Steps 2 to 5 together are exactly what the pipeline's `clean` does:

```python
df.equals(clean(raw, cfg["clean"]))
```

## 6. log1p on skewed columns

EDA A7 found skew well above 1 for PM2.5 and the wind columns. log1p(x) = ln(1 + x) turns "times bigger" into "plus
a fixed step", so the model judges errors as percentages and the rare spikes stop dominating the loss. The +1 keeps
zeros safe. Predictions are mapped back with expm1 before computing MAE, RMSE and R2. The transform itself is applied
inside the per-fold scaler (step 8); this cell confirms its effect on 2010 to 2013 (the test year stays unseen).

```python
cols = ["pm25", "ws", "ws_cum", "snow_cum", "rain_cum"]
dev = df[df.index.year < cfg["split"]["test_year"]]
pd.DataFrame({"raw skew": dev[cols].skew(), "log1p skew": np.log1p(dev[cols]).skew()}).round(2).T
```

## 7. Chronological folds

Fold k evaluates on one calendar year and trains on everything up to 24 hours before it. The 24-hour gap is
an embargo: with gap >= forecast horizon, no training target can fall inside the evaluation year.
2014 is the final test year and is not touched until the end.

```python
folds = make_folds(df.index, **cfg["split"])
pd.DataFrame([{
    "fold": f["name"],
    "train from": f["train"][0],
    "train until (excl.)": f["train"][1],
    "eval from": f["eval"][0],
    "eval until (excl.)": f["eval"][1],
    "train rows": fold_mask(df.index, f["train"]).sum(),
    "labelled eval rows": df.loc[fold_mask(df.index, f["eval"]), "pm25"].notna().sum(),
} for f in folds])
```

```python
viz.fold_timeline(folds);
```

## 8. Leak-free scaling

Each fold gets its own scaler, fitted on that fold's training rows only: log1p for the skewed columns, then a
z-score. Shown here for fold `cv1`.

```python
f = folds[0]
tr, ev = fold_mask(df.index, f["train"]), fold_mask(df.index, f["eval"])
sc = Scaler(**cfg["scale"]).fit(df[tr])
stats = pd.DataFrame({"mean": pd.Series(sc.mean), "std": pd.Series(sc.std)}).round(3)
stats.assign(log1p=stats.index.isin(sc.log))
```

On the evaluation year the scaled columns are close to, but not exactly, mean 0 and std 1. That is expected:
the statistics come from 2010 only, and nothing from 2011 was used to compute them.

```python
z = sc.transform(df[ev])
z[sc.cols].agg(["mean", "std"]).round(2)
```

```python
np.allclose(sc.inverse(z["pm25"], "pm25"), df.loc[ev, "pm25"], equal_nan=True)
```

## Summary

- 43,824 hourly rows, complete grid; only `pm25` is missing (4.72%), mostly in 2010 to 2012.
- `Iws` decoded into hourly wind speed; `Is` / `Ir` into snow / rain flags; wind direction one-hot.
- Label untouched, missing flag added, input copy filled causally for at most 3 hours.
- Expanding-window folds on 2011, 2012, 2013, test on 2014, 24-hour embargo.
- Scalers fitted on training rows only, saved per fold.

**What "ready for training" means here.** The output is a clean, validated hourly table plus fold boundaries and
per-fold scalers. Anything that learns a statistic from the data (scaler means, the hour-of-day climatology behind
one feature) must be fitted inside each fold's training years, so it cannot be computed once for the whole file
without leaking. So features are built in `03_phase1_models.ipynb`: lags and rolling statistics from past values
only, and the fold-dependent parts (the climatology anomaly and the target-hour calendar) per fold and horizon.
